Hashing in Python — Data Structures and Algorithms (DSA)
Hashing ni technique inayotumika kuhifadhi na kutafuta data kwa haraka kwa kutumia Hash Function.

Katika Python, hashing inatumika kwenye:
1. Hash Map (dict)
2. Hash Set (set)
3. Hash Table (implementation from scratch)

Kwa kawaida, operations za insertion, searching na deletion zina average time complexity ya O(1).

# 1. Hash Map Implementation Using Dictionary
Python dict ni implementation ya Hash Map.

```python
# Create Hash Map
hash_map = {}

# Insert
hash_map["name"] = "Emmanuel"
hash_map["age"] = 24
hash_map["country"] = "Tanzania"

# Access
print(hash_map["name"])

# Update
hash_map["age"] = 25

# Search
print("name" in hash_map)

# Delete
del hash_map["country"]

print(hash_map)
```

Hash Map Operations

| Operation | Python | Average Complexity |
|---|---|---|
| Insert | `hash_map[key] = value` | O(1) |
| Search | `key in hash_map` | O(1) |
| Access | `hash_map[key]` | O(1) |
| Delete | `del hash_map[key]` | O(1) |
| Update | `hash_map[key] = value` | O(1) |

# 2. Hash Set Implementation

Hash Set inahifadhi unique elements pekee.

```python
hash_set = set()

# Insert
hash_set.add(10)
hash_set.add(20)
hash_set.add(30)

# Duplicate
hash_set.add(10)

print(hash_set)

# Search
print(20 in hash_set)

# Remove
hash_set.remove(20)

print(hash_set)
```

| `discard()` | O(1) average |

#3. Implement Hash Table From Scratch
Sasa tuimplement Hash Table wenyewe bila kutumia Python dict.

Tutatumia:
- Array (Python list)
- Hash Function
- Separate Chaining kwa collision handling
Implementation

```python
class HashTable:

    def __init__(self, size=10):
        if size <= 0:
            raise ValueError("Size must be positive")

        self.size = size
        self.table = [[] for _ in range(size)]

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)

        bucket = self.table[index]

        for pair in bucket:
            if pair[0] == key:
                pair[1] = value
                return

        bucket.append([key, value])

    def search(self, key):
        index = self.hash_function(key)

        bucket = self.table[index]

        for pair in bucket:
            if pair[0] == key:
                return pair[1]

        raise KeyError(key)

    def delete(self, key):
        index = self.hash_function(key)

        bucket = self.table[index]

        for i, pair in enumerate(bucket):
            if pair[0] == key:
                bucket.pop(i)
                return

        raise KeyError(key)

    def display(self):
        for index, bucket in enumerate(self.table):
            print(index, ":", bucket)
```

```python
ht = HashTable()

# Insert
ht.insert("name", "Emmanuel")
ht.insert("age", 24)
ht.insert("country", "Tanzania")

# Search
print(ht.search("name"))

# Delete
ht.delete("age")

# Display
ht.display()
```

#4. Hash Function Inafanyaje Kazi?
Hash Function inabadilisha key kuwa hash value, ambayo tunaweza kuitumia kupata index kwenye Hash Table.

Mfano rahisi:

```python
def hash_function(key, size):
    return key % size


print(hash_function(25, 10))  # 5
print(hash_function(37, 10))  # 7
print(hash_function(42, 10))  # 2
```

42  ─────── 42 % 10 ────────── 2

#6. Hashing kwenye DSA Problems

## Example 1: Two Sum Using Hash Map
Problem: Una array na target. Tafuta indices za numbers mbili ambazo jumla yake ni target.

```python
def two_sum(nums, target):

    seen = {}

    for i, num in enumerate(nums):

        complement = target - num

        if complement in seen:
            return [seen[complement], i]

        seen[num] = i

    return []


nums = [2, 7, 11, 15]

print(two_sum(nums, 9))

#Time Complexity: O(n) average

#Space Complexity: O(n)
```

## Example 2: Find Duplicates Using Hash Set

```python
def contains_duplicate(nums):

    seen = set()

    for num in nums:

        if num in seen:
            return True

        seen.add(num)

    return False


nums = [1, 2, 3, 4, 2]

print(contains_duplicate(nums))
# True
#Time Complexity: O(n) average

#Space Complexity: O(n)
```

##  Example 3: Frequency Counter

```python
def count_frequency(nums):

    frequency = {}

    for num in nums:
        frequency[num] = frequency.get(num, 0) + 1

    return frequency


nums = [1, 2, 2, 3, 3, 3]

print(count_frequency(nums))
```

# 8. Hash Map vs Hash Set vs Hash Table

| Feature | Hash Map | Hash Set | Hash Table |
|---|---|---|---|
| Purpose | Key-value storage | Unique values | Hash-based storage |
| Python | `dict` | `set` | Custom implementation |
| Duplicates | Unique keys | No duplicates | Depends on implementation |
| Lookup | O(1) average | O(1) average | O(1) average |
| Common use | Frequency counting | Duplicate detection | Implementing maps/sets |
