#Linked List in Python — Data Structures and Algorithms (DSA)
A Linked List is a linear data structure consisting of a collection of elements called nodes, where each node stores data and a reference to another node.

Unlike arrays, linked list nodes do not need to occupy consecutive memory locations.
#1. Structure of a Linked List
Each node in a singly linked list contains two components:
- Data: The value stored in the node.
- Next: A reference to the next node.

       Head
         ↓
    ┌────┬────┐    ┌────┬────┐    ┌────┬────┐
    │ 10 │  ●─┼───→│ 20 │  ●─┼───→│ 30 │None│
    └────┴────┘    └────┴────┘    └────┴────┘
                                      ↑
                                     Tail

A simplified representation:
Head → 10 → 20 → 30 → None

| Terminology | Meaning |
|---|---|
| Node | Individual element in a linked list |
| Head | Reference to the first node |
| Tail | Reference to the last node |
| Next | Reference to the next node |
| Previous | Reference to the previous node in a doubly linked list |
| None | Indicates the end of a non-circular linked list |



#2. Array vs Linked List
Array

Index:   0     1     2     3
       ┌────┬─────┬─────┬─────┐
       │ 10 │ 20  │ 30  │ 40  │
       └────┴─────┴─────┴─────┘

Head → [10] → [20] → [30] → [40] → None

| Feature | Python List (Dynamic Array) | Linked List |
|---|---|---|
| Memory layout | Contiguous array of references | Separate linked nodes |
| Random access | O(1) | O(n) |
| Searching | O(n) | O(n) |
| Insert at beginning | O(n) | O(1) |
| Insert at end | O(1) amortized | O(1) with tail |
| Delete first element | O(n) | O(1) |
| Delete last element | O(1) | O(n) for singly linked list |
| Insert after a known node | O(n) worst case | O(1) |
| Memory overhead | Generally lower | Generally higher |
| Cache locality | Better | Generally worse |

Important: Linked list insertion and deletion are O(1) only when the necessary node references are already available. Finding the appropriate node can take O(n).


#3. Types of Linked Lists
##3.1 Singly Linked List
Each node contains data and a reference to the next node.

Head → 10 → 20 → 30 → None

Traversal is possible in one direction.


```python
class Node:

    def __init__(self, data):

        self.data = data
        self.next = None
```

##3.2 Doubly Linked List
Each node contains data, a reference to the previous node and a reference to the next node.

None ← 10 ⇄ 20 ⇄ 30 → None
       ↑          ↑
      Head       Tail

```python
class DoublyNode:

    def __init__(self, data):

        self.data = data
        self.next = None
        self.prev = None
```

##3.3 Circular Singly Linked List
The last node points back to the first node instead of None.

Head → 10 → 20 → 30
        ↑         |
        └─────────┘

Common applications include round-robin scheduling and circular buffers.
##3.4 Circular Doubly Linked List
Each node has both previous and next references, and the first and last nodes are connected.

       ┌─────────────────┐
       ↓                 ↑
       10 ⇄ 20 ⇄ 30 ⇄ 40
       ↑                 ↓
       └─────────────────┘


| Type | Forward traversal | Backward traversal | Circular |
|---|---|---|---|
| Singly | Yes | No | No |
| Doubly | Yes | Yes | No |
| Circular Singly | Yes | No | Yes |
| Circular Doubly | Yes | Yes | Yes |

#4. Implementing a Singly Linked List From Scratch
##4.1 Create a Node

```python
class Node:

    def __init__(self, data):

        self.data = data
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
```

Connecting the nodes:

```python
node1.next = node2

node2.next = node3

head = node1
```

Result:

Head → 10 → 20 → 30 → None

Accessing values:


```python
print(head.data)            # 10

print(head.next.data)       # 20

print(head.next.next.data)  # 30
```

##4.2 Create the Linked List Class

```python
class LinkedList:

    def __init__(self):

        self.head = None
        self.tail = None
        self.size = 0
```

Initially, the linked list is empty.

Head → None

Tail → None

Size = 0



#5. Linked List Operations
##5.1 Append — Insert at the End

Before:

Head → 10 → 20 → None

Append 30:

Head → 10 → 20 → 30 → None

Implementation:

```python
def append(self, data):

    new_node = Node(data)

    if self.head is None:

        self.head = new_node
        self.tail = new_node

    else:

        self.tail.next = new_node
        self.tail = new_node

    self.size += 1
```

Time complexity: O(1), because we maintain a tail reference.

Without a tail reference, appending would require traversing the list in O(n).
##5.2 Prepend — Insert at the Beginning

Before:

Head → 10 → 20 → 30 → None

Prepend 5:

Head → 5 → 10 → 20 → 30 → None

Implementation:

```python
def prepend(self, data):

    new_node = Node(data)

    new_node.next = self.head

    self.head = new_node

    if self.tail is None:

        self.tail = new_node

    self.size += 1
```
