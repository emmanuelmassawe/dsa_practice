#Linked List in Python — Data Structures and Algorithms (DSA)
A Linked List is a linear data structure consisting of a collection of elements called nodes, where each node stores data and a reference to another node.

Unlike arrays, linked list nodes do not need to occupy consecutive memory locations.
#1. Structure of a Linked List
Each node in a singly linked list contains two components:
- Data: The value stored in the node.
- Next: A reference to the next node.

       Head
         ↓
    ┌────┬────┐    ┌────┬────┐    ┌────┬────┐
    │ 10 │  ●─┼───→│ 20 │  ●─┼───→│ 30 │None│
    └────┴────┘    └────┴────┘    └────┴────┘
                                      ↑
                                     Tail

A simplified representation:
Head → 10 → 20 → 30 → None

| Terminology | Meaning |
|---|---|
| Node | Individual element in a linked list |
| Head | Reference to the first node |
| Tail | Reference to the last node |
| Next | Reference to the next node |
| Previous | Reference to the previous node in a doubly linked list |
| None | Indicates the end of a non-circular linked list |



#2. Array vs Linked List
Array

Index:   0     1     2     3
       ┌────┬─────┬─────┬─────┐
       │ 10 │ 20  │ 30  │ 40  │
       └────┴─────┴─────┴─────┘

Head → [10] → [20] → [30] → [40] → None

| Feature | Python List (Dynamic Array) | Linked List |
|---|---|---|
| Memory layout | Contiguous array of references | Separate linked nodes |
| Random access | O(1) | O(n) |
| Searching | O(n) | O(n) |
| Insert at beginning | O(n) | O(1) |
| Insert at end | O(1) amortized | O(1) with tail |
| Delete first element | O(n) | O(1) |
| Delete last element | O(1) | O(n) for singly linked list |
| Insert after a known node | O(n) worst case | O(1) |
| Memory overhead | Generally lower | Generally higher |
| Cache locality | Better | Generally worse |

Important: Linked list insertion and deletion are O(1) only when the necessary node references are already available. Finding the appropriate node can take O(n).


#3. Types of Linked Lists
##3.1 Singly Linked List
Each node contains data and a reference to the next node.

Head → 10 → 20 → 30 → None

Traversal is possible in one direction.


```python
class Node:

    def __init__(self, data):

        self.data = data
        self.next = None
```

##3.2 Doubly Linked List

       ↑          ↑
      Head       Tail

```python
class DoublyNode:

    def __init__(self, data):

        self.data = data
        self.next = None
        self.prev = None
```

##3.3 Circular Singly Linked List

Head → 10 → 20 → 30
        ↑         |

##3.4 Circular Doubly Linked List

#5. Linked List Operations
##5.1 Append — Insert at the End

Before:

Head → 10 → 20 → None

Append 30:

Head → 10 → 20 → 30 → None

Implementation:

```python
def append(self, data):

    new_node = Node(data)

    if self.head is None:

        self.head = new_node
        self.tail = new_node

    else:

        self.tail.next = new_node
        self.tail = new_node

    self.size += 1
```

Time complexity: O(1), because we maintain a tail reference.

Without a tail reference, appending would require traversing the list in O(n).
##5.2 Prepend — Insert at the Beginning

Before:

Head → 10 → 20 → 30 → None

Prepend 5:

Head → 5 → 10 → 20 → 30 → None

Implementation:

```python
def prepend(self, data):

    new_node = Node(data)

    new_node.next = self.head

    self.head = new_node

    if self.tail is None:

        self.tail = new_node

    self.size += 1
```

Time complexity: O(1).
##5.3 Traversal
Traversal means visiting every node in the linked list.

```python
def display(self):

    current = self.head

    while current is not None:

        print(current.data, end=" → ")

        current = current.next

    print("None")
```

Time complexity: O(n)
Auxiliary space: O(1)
##5.4 Searching
Find the index of a particular value.

```python
def search(self, target):

    current = self.head

    index = 0

    while current is not None:

        if current.data == target:

            return index

        current = current.next

        index += 1

    return -1
```

Time complexity: O(n)
Auxiliary space: O(1)

Linked lists generally use linear search because they do not support constant-time random access.
##5.5 Insert at a Specific Index

Before:

10 → 20 → 30 → None

Insert 15 at index 1:

10 → 15 → 20 → 30 → None



```python
def insert(self, index, data):

    if index < 0 or index > self.size:

        raise IndexError("Invalid index")

    if index == 0:

        self.prepend(data)
        return

    if index == self.size:

        self.append(data)
        return

    current = self.head

    for _ in range(index - 1):

        current = current.next

    new_node = Node(data)

    new_node.next = current.next

    current.next = new_node

    self.size += 1
```

The order of reference updates is important:

```python
#new_node.next = current.next

#current.next = new_node
```

Otherwise, you could lose the reference to the remaining nodes.

Time complexity: O(n) worst case.
##5.6 Delete the First Node

Before:

Head → 10 → 20 → 30 → None

After:

Head → 20 → 30 → None

```python
def pop_left(self):

    if self.head is None:

        raise IndexError("Linked list is empty")

    removed = self.head.data

    self.head = self.head.next

    self.size -= 1

    if self.head is None:

        self.tail = None

    return removed
```

Time complexity: O(1).
##5.7 Delete the Last Node

Before:

10 → 20 → 30 → None

After:

10 → 20 → None


```python
def pop(self):

    if self.head is None:

        raise IndexError("Linked list is empty")

    if self.head is self.tail:

        return self.pop_left()

    current = self.head

    while current.next is not self.tail:

        current = current.next

    removed = self.tail.data

    current.next = None

    self.tail = current

    self.size -= 1

    return removed
```

Time complexity: O(n).

Even with a tail reference, a singly linked list must find the second-to-last node because it cannot move backward.
##5.8 Delete by Value

Before:

10 → 20 → 30 → 40 → None

Delete 30:

10 → 20 → 40 → None



```python
def delete(self, target):

    current = self.head

    previous = None

    while current is not None:

        if current.data == target:

            if previous is None:

                self.pop_left()

            else:

                previous.next = current.next

                if current is self.tail:

                    self.tail = previous

                self.size -= 1

            return True

        previous = current

        current = current.next

    return False
```

This implementation deletes the first occurrence of the target.

Time complexity: O(n).
#6. Complete Singly Linked List Implementation
Combining the operations above:

```python
class Node:

    def __init__(self, data):

        self.data = data
        self.next = None


class LinkedList:

    def __init__(self):

        self.head = None
        self.tail = None
        self.size = 0

    def append(self, data):

        new_node = Node(data)

        if self.head is None:

            self.head = new_node
            self.tail = new_node

        else:

            self.tail.next = new_node
            self.tail = new_node

        self.size += 1

    def prepend(self, data):

        new_node = Node(data)

        new_node.next = self.head

        self.head = new_node

        if self.tail is None:

            self.tail = new_node

        self.size += 1

    def insert(self, index, data):

        if not 0 <= index <= self.size:

            raise IndexError("Invalid index")

        if index == 0:

            self.prepend(data)
            return

        if index == self.size:

            self.append(data)
            return

        current = self.head

        for _ in range(index - 1):

            current = current.next

        new_node = Node(data)

        new_node.next = current.next

        current.next = new_node

        self.size += 1

    def search(self, target):

        current = self.head

        index = 0

        while current is not None:

            if current.data == target:

                return index

            current = current.next

            index += 1

        return -1

    def pop_left(self):

        if self.head is None:

            raise IndexError("Linked list is empty")

        removed = self.head.data

        self.head = self.head.next

        self.size -= 1

        if self.head is None:

            self.tail = None

        return removed

    def pop(self):

        if self.head is None:

            raise IndexError("Linked list is empty")

        if self.head is self.tail:

            return self.pop_left()

        current = self.head

        while current.next is not self.tail:

            current = current.next

        removed = self.tail.data

        current.next = None

        self.tail = current

        self.size -= 1

        return removed

    def delete(self, target):

        current = self.head

        previous = None

        while current is not None:

            if current.data == target:

                if previous is None:

                    self.pop_left()

                else:

                    previous.next = current.next

                    if current is self.tail:

                        self.tail = previous

                    self.size -= 1

                return True

            previous = current

            current = current.next

        return False

    def display(self):

        current = self.head

        while current is not None:

            print(current.data, end=" → ")

            current = current.next

        print("None")

    def __len__(self):

        return self.size

    def is_empty(self):

        return self.size == 0
```

#7. Doubly Linked List Implementation
A doubly linked list maintains both next and prev references.

This allows efficient insertion and deletion at both ends.

```python
class DoublyNode:

    def __init__(self, data):

        self.data = data
        self.next = None
        self.prev = None


class DoublyLinkedList:

    def __init__(self):

        self.head = None
        self.tail = None
        self.size = 0

    def append(self, data):

        new_node = DoublyNode(data)

        if self.head is None:

            self.head = new_node
            self.tail = new_node

        else:

            new_node.prev = self.tail

            self.tail.next = new_node

            self.tail = new_node

        self.size += 1

    def prepend(self, data):

        new_node = DoublyNode(data)

        if self.head is None:

            self.head = new_node
            self.tail = new_node

        else:

            new_node.next = self.head

            self.head.prev = new_node

            self.head = new_node

        self.size += 1

    def pop(self):

        if self.tail is None:

            raise IndexError("Linked list is empty")

        removed = self.tail.data

        self.tail = self.tail.prev

        if self.tail is None:

            self.head = None

        else:

            self.tail.next = None

        self.size -= 1

        return removed

    def pop_left(self):

        if self.head is None:

            raise IndexError("Linked list is empty")

        removed = self.head.data

        self.head = self.head.next

        if self.head is None:

            self.tail = None

        else:

            self.head.prev = None

        self.size -= 1

        return removed

    def display_forward(self):

        current = self.head

        while current:

            print(current.data, end=" → ")

            current = current.next

        print("None")

    def display_backward(self):

        current = self.tail

        while current:

            print(current.data, end=" → ")

            current = current.prev

        print("None")
```

```python
dll = DoublyLinkedList()

dll.append(10)
dll.append(20)
dll.append(30)

dll.display_forward()

dll.display_backward()

print(dll.pop())
```

Singly vs Doubly Linked List

| Operation | Singly (with tail) | Doubly (with tail) |
|---|---|---|
| Append | O(1) | O(1) |
| Prepend | O(1) | O(1) |
| Delete first | O(1) | O(1) |
| Delete last | O(n) | O(1) |
| Search | O(n) | O(n) |
| Delete known node | O(n) generally | O(1) |
| Backward traversal | No | Yes |
| References per node | 1 | 2 |



#8. Circular Linked List Implementation
In a circular linked list, the tail points back to the head.

```python
class CircularLinkedList:

    def __init__(self):

        self.head = None
        self.tail = None

    def append(self, data):

        new_node = Node(data)

        if self.head is None:

            self.head = new_node
            self.tail = new_node

            new_node.next = new_node

        else:

            new_node.next = self.head

            self.tail.next = new_node

            self.tail = new_node

    def display(self):

        if self.head is None:
            return

        current = self.head

        while True:

            print(current.data, end=" → ")

            current = current.next

            if current is self.head:
                break

        print("Head")
```

```python
cll = CircularLinkedList()

cll.append(10)
cll.append(20)
cll.append(30)

cll.display()
```

Important: You cannot use while current is not None to terminate circular linked list traversal because the last node points back to the first node.
