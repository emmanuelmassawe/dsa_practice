A Tree is a non-linear data structure that stores data in a hierarchical structure using nodes connected by edges.

Trees are important in software engineering, databases, search algorithms, compilers, and machine learning.

#3. Binary Tree Implementation in Python
A Binary Tree is a tree in which each node has at most two children:
- Left child
- Right child
Example

         10
        /  \
       5    20
      / \
     3   7

Step 1: Create a Node Class

```python
class Node:
  def __init__(self, value):
    self.value = value
    self.left = None
    self.right = None

```

Each node contains:
- value: Data stored in the node.
- left: Reference to the left child.
- right: Reference to the right child.

```python
root = Node(10)
root.left = Node(5)
root.right = Node(20)
root.left.left = Node(3)
root.left.right = Node(7)
```

```python
print(root.value)
print(root.left.value)
print(root.right.value)
print(root.left.left.value)
print(root.left.right.value)
```

Time Complexity: O(1) for accessing a node through a fixed number of references.

However, searching for an arbitrary value in a general binary tree takes O(n) in the worst case.

#4. Tree Traversal Algorithms
Tree traversal means visiting every node in a tree.

There are two major approaches:
1. Depth-First Search (DFS)
2. Breadth-First Search (BFS)
4.1 Depth-First Search (DFS)

### DFS explores one branch deeply before moving to another branch.

There are three common DFS traversal orders:
- Inorder
- Preorder
- Postorder
Consider:

A. Inorder Traversal
Left → Root → Right

```python
def inorder(root):

    if root is None:
        return

    inorder(root.left)

    print(root.value)

    inorder(root.right)


inorder(root)
```

Time Complexity: O(n)

Auxiliary Space Complexity: O(h)

Where h is the height of the tree, measured by the recursion stack depth up to a constant difference.

Inorder traversal produces sorted values when applied to a valid Binary Search Tree.

B. Preorder Traversal
Root → Left → Right

```python
def preorder(root):

    if root is None:
        return

    print(root.value)

    preorder(root.left)

    preorder(root.right)


preorder(root)
```

C. Postorder Traversal
Left → Right → Root

```python
def postorder(root):

    if root is None:
        return

    postorder(root.left)

    postorder(root.right)

    print(root.value)


postorder(root)
```

Postorder is useful when processing children before their parent, such as deleting tree structures or evaluating expression trees.

#5. Breadth-First Search (BFS)
BFS visits nodes level by level.

It uses a Queue.

Example:

```python
from collections import deque

def bfs(root):

    if root is None:
        return

    queue = deque([root])

    while queue:

        node = queue.popleft()

        print(node.value)

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)


bfs(root)
```

Time Complexity: O(n)

Space Complexity: O(w)

Where w is the maximum number of nodes stored in the queue at one level, bounded by O(n).

DFS vs BFS

| Feature | DFS | BFS |
|---|---|---|
| Data Structure | Stack / Recursion | Queue |
| Exploration | Depth first | Level by level |
| Time Complexity | O(n) | O(n) |
| Space Complexity | O(h) | O(w) |
| Common Use | Tree recursion | Level-order traversal |

#6. Binary Search Tree (BST)
A Binary Search Tree is a special binary tree that maintains an ordering property.

For every node:
- All values in the left subtree are smaller than the node's value.
- All values in the right subtree are greater than the node's value.
Here, we'll assume duplicate values are not allowed.

Example:

```python
class Node:

    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


class BinarySearchTree:

    def __init__(self):
        self.root = None

    def insert(self, value):

        if self.root is None:
            self.root = Node(value)
            return

        current = self.root

        while True:

            if value < current.value:

                if current.left is None:
                    current.left = Node(value)
                    return

                current = current.left

            elif value > current.value:

                if current.right is None:
                    current.right = Node(value)
                    return

                current = current.right

            else:
                return

    def search(self, value):

        current = self.root

        while current:

            if value == current.value:
                return True

            elif value < current.value:
                current = current.left

            else:
                current = current.right

        return False
```

```python
bst = BinarySearchTree()

bst.insert(50)
bst.insert(30)
bst.insert(70)
bst.insert(20)
bst.insert(40)
bst.insert(60)
bst.insert(80)

print(bst.search(40))  # True
print(bst.search(90))  # False
```

How BST Searching Works
Suppose we want to find 60.

             50
            /  \
          30    70
               /  \
              60   80

Steps:
1. Start at 50.
2. Since 60 > 50, move right.
3. At 70, since 60 < 70, move left.
4. Find 60.

Instead of checking every node, BST uses comparisons to eliminate entire subtrees.
BST Time Complexity

| Operation | Balanced BST | Skewed BST |
|---|---|---|
| Search | O(log n) | O(n) |
| Insert | O(log n) | O(n) |
| Delete | O(log n) | O(n) |
| Inorder Traversal | O(n) | O(n) |

A normal BST does not automatically stay balanced.

For example:

10
  \
   20
     \
      30
        \
         40


This is a skewed tree, and searching may take O(n) time.

Self-balancing trees such as AVL and Red-Black Trees address this problem
