# 2. Bubble Sort
Bubble Sort repeatedly compares adjacent elements and swaps them if they are in the wrong order.

After each complete pass, the largest remaining element moves to its correct position.
How Bubble Sort Works

Initial: [5, 3, 8, 1]

Compare 5 and 3 → Swap
[3, 5, 8, 1]

Compare 5 and 8 → No swap
[3, 5, 8, 1]

Compare 8 and 1 → Swap
[3, 5, 1, 8]

First pass completed.
8 is now in its correct position.



```python
def bubble_sort(arr):

    n = len(arr)

    for i in range(n - 1):

        swapped = False

        for j in range(n - i - 1):

            if arr[j] > arr[j + 1]:

                arr[j], arr[j + 1] = arr[j + 1], arr[j]

                swapped = True

        if not swapped:
            break

    return arr


arr = [50, 20, 40, 10, 30]

print(bubble_sort(arr))
```

| Property | Value |
|---|---|
| Best case | O(n) |
| Average case | O(n²) |
| Worst case | O(n²) |
| Space complexity | O(1) |
| Stable | Yes |
| In-place | Yes |



#3. Selection Sort
Selection Sort repeatedly finds the smallest element in the unsorted portion and moves it to its correct position.
How Selection Sort Works

Initial: [64, 25, 12, 22, 11]

Pass 1: Find minimum = 11
[11, 25, 12, 22, 64]

Pass 2: Find minimum = 12
[11, 12, 25, 22, 64]

Pass 3: Find minimum = 22
[11, 12, 22, 25, 64]

Pass 4: Find minimum = 25
[11, 12, 22, 25, 64]



```python
def selection_sort(arr):

    n = len(arr)

    for i in range(n - 1):

        min_index = i

        for j in range(i + 1, n):

            if arr[j] < arr[min_index]:

                min_index = j

        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr


arr = [64, 25, 12, 22, 11]

print(selection_sort(arr))
```

| Property | Value |
|---|---|
| Best case | O(n²) |
| Average case | O(n²) |
| Worst case | O(n²) |
| Space complexity | O(1) |
| Stable | No |
| In-place | Yes |

Selection Sort performs relatively few swaps, but it still performs many comparisons, even when the array is already sorted.
#4. Insertion Sort
Insertion Sort builds a sorted portion of the array by inserting each element into its correct position.

It works similarly to arranging playing cards in your hand.
How Insertion Sort Works

```python
def insertion_sort(arr):

    for i in range(1, len(arr)):

        key = arr[i]

        j = i - 1

        while j >= 0 and arr[j] > key:

            arr[j + 1] = arr[j]

            j -= 1

        arr[j + 1] = key

    return arr


arr = [5, 3, 4, 1]

print(insertion_sort(arr))
```

| Property | Value |
|---|---|
| Best case | O(n) |
| Average case | O(n²) |
| Worst case | O(n²) |
| Space complexity | O(1) |
| Stable | Yes |
| In-place | Yes |

Insertion Sort is particularly efficient for small or nearly sorted arrays.
#5. Merge Sort
Merge Sort is a sorting algorithm based on Divide and Conquer and recursion.

It divides an array into smaller subarrays, sorts them independently, and merges them into a sorted array.
How Merge Sort Works

              [8, 3, 5, 1]
               /        \
           [8, 3]      [5, 1]
           /   \        /   \
         [8]   [3]    [5]   [1]

              MERGING

           [3, 8]      [1, 5]
               \        /
              [1, 3, 5, 8]

Merge Sort has three main steps:
1. Divide: Split the array into two halves.
2. Conquer: Recursively sort each half.
3. Combine: Merge the sorted halves.
Python Implementation

```python
def merge_sort(arr):

    # Base case
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])

    right = merge_sort(arr[mid:])

    return merge(left, right)


def merge(left, right):

    result = []

    i = 0
    j = 0

    while i < len(left) and j < len(right):

        if left[i] <= right[j]:

            result.append(left[i])
            i += 1

        else:

            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


arr = [8, 3, 5, 1, 9, 2]

print(merge_sort(arr))
```

Understanding Merge Sort Complexity
At each level, Merge Sort processes n elements.

The array is repeatedly divided in half, producing approximately log n levels.
\[
T(n) = 2T(n/2) + O(n)
\]
Therefore:
\[
T(n) = O(n\log n)
\]

| Property | Value |
|---|---|
| Best case | O(n log n) |
| Average case | O(n log n) |
| Worst case | O(n log n) |
| Space complexity | O(n) |
| Stable | Yes |
| In-place | No |

Merge Sort is useful for large datasets, external sorting, and sorting linked lists.
#6. Quick Sort
Quick Sort is another Divide and Conquer algorithm that uses an element called a pivot.

It partitions the array so that smaller elements move to one side of the pivot and larger elements move to the other side.
How Quick Sort Works

Array: [8, 3, 1, 7, 5]

Pivot = 5

Smaller      Pivot      Greater

[3, 1]        [5]        [8, 7]

Recursively sort both sides:

[1, 3]        [5]        [7, 8]

Final:

[1, 3, 5, 7, 8]

```python
def quick_sort(arr):

    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]

    middle = [x for x in arr if x == pivot]

    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)


arr = [8, 3, 1, 7, 5]

print(quick_sort(arr))
```

#11. Sorting Using Python Built-in Methods
Python provides two primary sorting methods:

```python
#sorted()
#list.sort()
```

##11.1 Using sorted()
Returns a new sorted list without modifying the original.

```python
arr = [50, 20, 40, 10, 30]

result = sorted(arr)

print(result)
print(arr)
```

```python
result = sorted(arr, reverse=True)

print(result)
# [50, 40, 30, 20, 10]
```

##11.2 Using list.sort()
Modifies the original list directly.

```python
arr = [50, 20, 40, 10, 30]

arr.sort()

print(arr)
# [10, 20, 30, 40, 50]
```

```python
arr.sort(reverse=True)
```

sorted() vs sort()

| Feature | `sorted()` | `list.sort()` |
|---|---|---|
| Returns | New sorted list | None |
| Modifies original | No | Yes |
| Supports tuples | Yes | No |
| Supports sets | Yes | No |
| Supports strings | Yes | No |
| Supports lists | Yes | Yes |
| Stable | Yes | Yes |
| Worst-case time | O(n log n) | O(n log n) |
