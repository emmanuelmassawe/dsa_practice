```python
# creating a list
arr = [2,4,6,4,7,8,9]

print(arr)
```

```python
#function that prints out a list
def print_item(arr):
  for x in arr:
    print(x)

print_item(arr)
```

####finding maxmum values

```python
def maximum_value(arr):
  maximum = arr[0]

  for i in arr:
    if i > maximum:
      maximum = i

  return maximum

maximum_value(arr)
```

finding minimum values

```python
def minimum_value(arr):
  minimum = arr[0]

  for i in arr:
    if i < minimum:
      minimum = i

  return minimum

minimum_value(arr)
```
