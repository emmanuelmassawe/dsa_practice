#Searching Algorithms in Python — Linear Search and Binary Search
Searching is the process of finding a particular element (target) within a data structure, such as an array or list.

There are two fundamental searching algorithms in DSA:
- Linear Search: Checks elements sequentially.
- Binary Search: Repeatedly divides a sorted search space into two halves.
Let's implement both from scratch and using Python's built-in methods and standard libraries.
#1. Linear Search
Linear Search checks every element one by one until it finds the target or reaches the end of the array.

The array does not need to be sorted.
How Linear Search Works

Array:  [30, 10, 50, 20, 40]
Index:    0   1   2   3   4

Target = 20

Step 1: 30 == 20? False
Step 2: 10 == 20? False
Step 3: 50 == 20? False
Step 4: 20 == 20? True

Return index 3

# 1.1 Implementation From Scratch

```python
def linear_search(arr, target):

    for i in range(len(arr)):

        if arr[i] == target:
            return i

    return -1


arr = [30, 10, 50, 20, 40]

print(linear_search(arr, 20))
```

If the target does not exist, the function returns -1.

#1.2 Using enumerate()
A more Pythonic implementation:

```python
def linear_search(arr, target):

    for index, value in enumerate(arr):

        if value == target:
            return index

    return -1


print(linear_search([30, 10, 50, 20, 40], 20))
# 3
```

##1.3 Linear Search Using Python Built-in Methods

Python provides the index() method, which performs a linear search on a list.

```python
arr = [30, 10, 50, 20, 40]

index = arr.index(20)

print(index)
# 3
```

However, if the element does not exist, Python raises a ValueError.

You can handle this using:

```python
def linear_search(arr, target):

    try:
        return arr.index(target)

    except ValueError:
        return -1
```

If you only want to check whether an element exists, use the in operator.

```python
arr = [30, 10, 50, 20, 40]

print(20 in arr)  # True
print(90 in arr)  # False
```

##1.5 Linear Search Complexity

| Scenario | Time Complexity |
|---|---|
| Best case | O(1) |
| Average case | O(n) |
| Worst case | O(n) |
| Auxiliary space (iterative) | O(1) |

Advantages: Simple to implement, works on unsorted arrays, and requires no preprocessing.

Disadvantages: Slow for large datasets because it may need to check every element

# 2. Binary Search
Binary Search is an efficient searching algorithm that repeatedly divides the search space into two halves.

The most important requirement is that the array must be sorted.
##2.1 How Binary Search Works
Consider:

Array: [10, 20, 30, 40, 50, 60, 70]
Index:   0   1   2   3   4   5   6

Target = 60

Step 1: Find the middle element

[10, 20, 30, 40, 50, 60, 70]
 ↑           ↑           ↑
left        mid        right

```python
#mid = (left + right) // 2
```

The middle element is 40.

Since 60 > 40, discard the left half, including the middle element.

Step 2: Search the right half

[50, 60, 70]
 ↑   ↑   ↑
left mid right

The middle element is 60.

Target found. Return index 5 from the original array.

Instead of checking six elements, Binary Search finds the target using only two comparisons with middle elements.

##2.2 Iterative Binary Search From Scratch

```python
def binary_search(arr, target):

    left = 0
    right = len(arr) - 1

    while left <= right:

        mid = (left + right) // 2

        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return -1


arr = [10, 20, 30, 40, 50, 60, 70]

print(binary_search(arr, 60))
# 5
```

Understanding the Conditions

| Condition | Action |
|---|---|
| `arr[mid] == target` | Return the middle index |
| `arr[mid] < target` | Search the right half |
| `arr[mid] > target` | Search the left half |
| `left > right` | Target does not exist |

##2.3 Recursive Binary Search
Since you have already learned recursion, you can also implement Binary Search recursively.

```python
def binary_search(arr, target, left, right):

    # Base case
    if left > right:
        return -1

    mid = (left + right) // 2

    if arr[mid] == target:
        return mid

    elif arr[mid] < target:

        return binary_search(
            arr, target, mid + 1, right
        )

    else:

        return binary_search(
            arr, target, left, mid - 1
        )


arr = [10, 20, 30, 40, 50, 60, 70]

print(binary_search(arr, 60, 0, len(arr) - 1))
# 5
```

| Implementation | Time Complexity | Auxiliary Space |
|---|---|---|
| Iterative | O(log n) | O(1) |
| Recursive | O(log n) | O(log n) |

The iterative implementation is generally preferable in Python because it avoids recursive function call overhead.

#3. Binary Search Using Python's bisect Library
Python provides a standard library called bisect, which implements binary-search-based operations on sorted sequences.

You do not need to install it.

```python
import bisect
```

Important Methods

| Method | Description | Time Complexity |
|---|---|---|
| `bisect_left()` | Find the leftmost insertion position | O(log n) |
| `bisect_right()` | Find the rightmost insertion position | O(log n) |
| `bisect()` | Alias for `bisect_right()` | O(log n) |
| `insort_left()` | Insert before equal elements | O(n) |
| `insort_right()` | Insert after equal elements | O(n) |
| `insort()` | Alias for `insort_right()` | O(n) |

## 3.1 Using bisect_left()



```python
from bisect import bisect_left

arr = [10, 20, 30, 40, 50, 60]

index = bisect_left(arr, 40)

print(index)
# 3
```

However, bisect_left() returns an insertion position, not necessarily the index of an existing element.


```python
print(bisect_left(arr, 35))
# 3
```

Even though 35 does not exist in the array, it returns the position where 35 should be inserted.

Correct Binary Search Implementation Using bisect

```python
from bisect import bisect_left


def binary_search(arr, target):

    index = bisect_left(arr, target)

    if index < len(arr) and arr[index] == target:
        return index

    return -1


arr = [10, 20, 30, 40, 50, 60]

print(binary_search(arr, 40))  # 3
print(binary_search(arr, 35))  # -1
```

Time Complexity: O(log n)

Auxiliary Space: O(1)
##3.2 Using bisect_right()
bisect_right() returns the insertion position after all occurrences of the target.

```python
from bisect import bisect_right

arr = [10, 20, 20, 20, 30, 40]

print(bisect_right(arr, 20))
# 4
```

## 3.3 Difference Between bisect_left() and bisect_right()



```python
from bisect import bisect_left, bisect_right

arr = [10, 20, 20, 20, 30, 40]

print(bisect_left(arr, 20))   # 1
print(bisect_right(arr, 20))  # 4
```

Array: [10, 20, 20, 20, 30, 40]
Index:   0   1   2   3   4   5
             ↑           ↑
        bisect_left   bisect_right

You can use these two functions to count how many times a target appears in a sorted array.

```python
def count_occurrences(arr, target):

    left = bisect_left(arr, target)
    right = bisect_right(arr, target)

    return right - left


print(count_occurrences(arr, 20))
# 3
```

Time Complexity: O(log n)
## 3.4 Using insort()
insort() inserts an element while maintaining the sorted order of the array.

```python
from bisect import insort

arr = [10, 20, 40, 50]

insort(arr, 30)

print(arr)
# [10, 20, 30, 40, 50]
```

Time Complexity: O(n)

Although finding the insertion position takes O(log n), shifting array elements takes O(n).

#4. Binary Search Using NumPy
NumPy provides searchsorted(), which performs binary search to find insertion positions.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

index = np.searchsorted(arr, 30)

print(index)
# 2
```

Like bisect, you must verify whether the target actually exists.

```python
def binary_search(arr, target):

    index = np.searchsorted(arr, target)

    if index < len(arr) and arr[index] == target:
        return int(index)

    return -1
```

NumPy also supports searching for multiple targets at once.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

targets = np.array([20, 40, 50])

print(np.searchsorted(arr, targets))
# [1 3 4]
```

This is particularly useful when working with numerical data and ML pipelines.

| Implementation | Simple | Moderate |
