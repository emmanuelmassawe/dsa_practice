#Graphs in Python — Data Structures and Algorithms (DSA)
A Graph is a non-linear data structure consisting of a collection of vertices (nodes) connected by edges.

Unlike trees, graphs do not necessarily follow a hierarchical structure. Nodes can connect to multiple other nodes, form cycles, and have multiple paths between them.

Graphs are widely used in social networks, recommendation systems, computer networks, AI agents, navigation systems, and knowledge graphs.

#1. Understanding Graph Structure
Consider the following graph:

       A -------- B
       |          |
       |          |
       C -------- D
        \        /
         \      /
            E

This graph contains:
- Vertices (V): A, B, C, D, E.
- Edges (E): A–B, A–C, B–D, C–D, C–E, D–E.

\[

# 4. Implement Graph From Scratch
Let's implement a graph using a Python class and an adjacency list.

```python
class Graph:

    def __init__(self, directed=False):
        self.graph = {}
        self.directed = directed

    def add_vertex(self, vertex):

        if vertex not in self.graph:
            self.graph[vertex] = []

    def add_edge(self, source, destination):

        self.add_vertex(source)
        self.add_vertex(destination)

        self.graph[source].append(destination)

        if not self.directed:
            self.graph[destination].append(source)

    def remove_edge(self, source, destination):

        if source in self.graph:
            if destination in self.graph[source]:
                self.graph[source].remove(destination)

        if not self.directed:
            if destination in self.graph:
                if source in self.graph[destination]:
                    self.graph[destination].remove(source)

    def get_neighbors(self, vertex):

        return self.graph.get(vertex, [])

    def display(self):

        for vertex, neighbors in self.graph.items():
            print(vertex, "->", neighbors)
```

```python
g = Graph()

g.add_edge("A", "B")
g.add_edge("A", "C")
g.add_edge("B", "D")
g.add_edge("C", "D")

g.display()
```

To create a directed graph:

```python
g = Graph(directed=True)

g.add_edge("A", "B")
g.add_edge("B", "C")

g.display()
```

# 5. Graph Traversal Algorithms
Graph traversal is the process of visiting nodes in a graph.

The two fundamental graph traversal algorithms are:
- Breadth-First Search (BFS) — uses a Queue.
- Depth-First Search (DFS) — uses a Stack or Recursion.
Consider this graph for both algorithms:

```python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "E"],
    "D": ["B", "F"],
    "E": ["C", "F"],
    "F": ["D", "E"]
}
```

## 5.1 Breadth-First Search (BFS)
BFS explores nodes level by level before moving deeper into the graph.

It uses the Queue (FIFO) data structure that we discussed earlier.
Implementation

```python
from collections import deque

def bfs(graph, start):

    visited = {start}
    queue = deque([start])
    result = []

    while queue:

        node = queue.popleft()

        result.append(node)

        for neighbor in graph[node]:

            if neighbor not in visited:

                visited.add(neighbor)
                queue.append(neighbor)

    return result


print(bfs(graph, "A"))
```

Step-by-step execution

| Step | Current Node | Queue | Visited Order |
|---|---|---|---|
| Initial | — | [A] | — |
| 1 | A | [B, C] | A |
| 2 | B | [C, D] | A, B |
| 3 | C | [D, E] | A, B, C |
| 4 | D | [E, F] | A, B, C, D |
| 5 | E | [F] | A, B, C, D, E |
| 6 | F | [] | A, B, C, D, E, F |

Time Complexity: O(V + E)

Space Complexity: O(V)

BFS is particularly useful for finding the shortest path in an unweighted graph.
##5.2 Depth-First Search (DFS)
DFS explores a branch as deeply as possible before backtracking.
Recursive Implementation

```python
def dfs(graph, start):

    visited = set()
    result = []

    def traverse(node):

        visited.add(node)
        result.append(node)

        for neighbor in graph[node]:

            if neighbor not in visited:
                traverse(neighbor)

    traverse(start)

    return result


print(dfs(graph, "A"))
```

Iterative DFS Using Stack
Remember that Python list can be used as a Stack.

```python
def dfs_iterative(graph, start):

    visited = set()
    stack = [start]
    result = []

    while stack:

        node = stack.pop()

        if node in visited:
            continue

        visited.add(node)
        result.append(node)

        for neighbor in reversed(graph[node]):

            if neighbor not in visited:
                stack.append(neighbor)

    return result


print(dfs_iterative(graph, "A"))
```

Time Complexity: O(V + E)

Space Complexity: Recursive DFS uses O(V) worst-case auxiliary space.
The iterative implementation above can use O(V + E) space because duplicate pending nodes may enter the stack.

BFS vs DFS

| Feature | BFS | DFS |
|---|---|---|
| Data structure | Queue | Stack / Recursion |
| Principle | FIFO | LIFO |
| Traversal | Level by level | Depth first |
| Time complexity | O(V + E) | O(V + E) |
| Shortest unweighted path | Yes | Not guaranteed |
| Cycle detection | Supported | Supported |
| Connected components | Supported | Supported |
| Topological sorting | Kahn's algorithm | DFS-based algorithm |

Why do we need a visited set?

Unlike trees, graphs can contain cycles.

A → B → C → A → B → C → ...

Without tracking visited nodes, traversal could continue indefinitely.

#6. Shortest Path Algorithms
Shortest path algorithms find the most efficient route between nodes.
##6.1 Shortest Path Using BFS
Use BFS when all edges have equal cost.

```python
from collections import deque

def shortest_path(graph, start, target):

    queue = deque([start])
    parent = {start: None}

    while queue:

        node = queue.popleft()

        if node == target:

            path = []

            while node is not None:
                path.append(node)
                node = parent[node]

            return path[::-1]

        for neighbor in graph[node]:

            if neighbor not in parent:

                parent[neighbor] = node
                queue.append(neighbor)

    return None
```

```python
print(shortest_path(graph, "A", "F"))
```

Time Complexity: O(V + E)

Space Complexity: O(V)
## 6.2 Dijkstra's Algorithm
Dijkstra's algorithm finds the shortest path in a weighted graph with non-negative edge weights.

It uses a Min Heap (Priority Queue).

```python
import heapq

def dijkstra(graph, start):

    distances = {
        node: float("inf")
        for node in graph
    }

    distances[start] = 0

    heap = [(0, start)]

    while heap:

        current_distance, node = heapq.heappop(heap)

        if current_distance > distances[node]:
            continue

        for neighbor, weight in graph[node]:

            distance = current_distance + weight

            if distance < distances[neighbor]:

                distances[neighbor] = distance

                heapq.heappush(
                    heap,
                    (distance, neighbor)
                )

    return distances
```

```python
weighted_graph = {
    "A": [("B", 4), ("C", 1)],
    "B": [("A", 4), ("D", 1)],
    "C": [("A", 1), ("D", 5)],
    "D": [("B", 1), ("C", 5)]
}

print(dijkstra(weighted_graph, "A"))
```

Time Complexity: O((V + E) log V) for a simple graph.

Space Complexity: O(V + E) worst case for this heap-based implementation.

#7. Cycle Detection
Cycle detection determines whether a graph contains a cycle.
Cycle Detection in an Undirected Graph
We use DFS and track the parent of each node.

```python
def has_cycle(graph):

    visited = set()

    def dfs(node, parent):

        visited.add(node)

        for neighbor in graph[node]:

            if neighbor not in visited:

                if dfs(neighbor, node):
                    return True

            elif neighbor != parent:

                return True

        return False

    for node in graph:

        if node not in visited:

            if dfs(node, None):
                return True

    return False


graph = {
    "A": ["B", "C"],
    "B": ["A", "C"],
    "C": ["A", "B"]
}

print(has_cycle(graph))
# True
```

Time Complexity: O(V + E)

Space Complexity: O(V)

This implementation assumes a simple undirected graph. Directed graphs require a different approach, such as tracking the active recursion stack or using topological sorting.

#8. Topological Sorting
Topological sorting arranges nodes in dependency order.

It only works on a Directed Acyclic Graph (DAG).

Consider an ML pipeline:

Data Ingestion
       ↓
Preprocessing
       ↓
Model Training
       ↓
Model Evaluation
       ↓
Deployment

A model cannot be evaluated before it has been trained.
Kahn's Algorithm
Kahn's algorithm uses a Queue and the in-degree of each node.

In-degree is the number of incoming edges.



```python
from collections import deque

def topological_sort(graph):

    indegree = {node: 0 for node in graph}

    for node in graph:

        for neighbor in graph[node]:
            indegree[neighbor] += 1

    queue = deque([
        node for node in indegree
        if indegree[node] == 0
    ])

    result = []

    while queue:

        node = queue.popleft()
        result.append(node)

        for neighbor in graph[node]:

            indegree[neighbor] -= 1

            if indegree[neighbor] == 0:
                queue.append(neighbor)

    if len(result) != len(graph):
        raise ValueError("Graph contains a cycle")

    return result
```

```python
pipeline = {
    "ingestion": ["preprocessing"],
    "preprocessing": ["training"],
    "training": ["evaluation"],
    "evaluation": ["deployment"],
    "deployment": []
}

print(topological_sort(pipeline))
```

Time Complexity: O(V + E)

Space Complexity: O(V)

This algorithm is useful in workflow orchestration, build systems, package dependency resolution and task scheduling.
#9. Connected Components
A connected component is a group of nodes connected to each other but disconnected from other groups in an undirected graph.

Component 1       Component 2      Component 3

A ----- B         D ----- E             F
|
C


```python
def count_components(graph):

    visited = set()
    count = 0

    def dfs(node):

        visited.add(node)

        for neighbor in graph[node]:

            if neighbor not in visited:
                dfs(neighbor)

    for node in graph:

        if node not in visited:

            dfs(node)
            count += 1

    return count
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["A"],
    "C": ["A"],
    "D": ["E"],
    "E": ["D"],
    "F": []
}

print(count_components(graph))
# 3
```

Time Complexity: O(V + E)

Space Complexity: O(V)
#10. Union-Find (Disjoint Set Union)
Union-Find, also called DSU, is a data structure used to determine whether two elements belong to the same connected group.

It has two main operations:

| Operation | Purpose |
|---|---|
| `find()` | Find the representative of a group |
| `union()` | Merge two groups |
| `connected()` | Check whether two elements belong to the same group |

```python
class UnionFind:

    def __init__(self, n):

        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, x):

        if self.parent[x] != x:

            self.parent[x] = self.find(
                self.parent[x]
            )

        return self.parent[x]

    def union(self, x, y):

        root_x = self.find(x)
        root_y = self.find(y)

        if root_x == root_y:
            return False

        if self.rank[root_x] < self.rank[root_y]:

            self.parent[root_x] = root_y

        elif self.rank[root_x] > self.rank[root_y]:

            self.parent[root_y] = root_x

        else:

            self.parent[root_y] = root_x
            self.rank[root_x] += 1

        return True

    def connected(self, x, y):

        return self.find(x) == self.find(y)
```

```python
uf = UnionFind(5)

uf.union(0, 1)
uf.union(1, 2)
uf.union(3, 4)

print(uf.connected(0, 2))  # True
print(uf.connected(0, 4))  # False
```

Before:

{0} {1} {2} {3} {4}

After:

{0, 1, 2}    {3, 4}

Time Complexity: O(α(n)) amortized per operation

With path compression and union by rank, this is practically close to constant time.

Union-Find is commonly used in cycle detection, network connectivity and Kruskal's algorithm.
#11. Minimum Spanning Tree (MST)
A Minimum Spanning Tree connects all vertices of a connected, weighted, undirected graph using the minimum possible total edge weight without creating cycles.

Consider:

          2
     A ------- B
     |         |
   1 |         | 3
     |         |
     C ------- D
          5


                    2
     A ------- B
     |         |
   1 |         | 3
     |         |
     C ------- D
          5

Total cost:
\[
1 + 2 + 3 = 6
\]
An MST always contains V - 1 edges.

The two main MST algorithms are:

| Algorithm | Data Structure | Time Complexity |
|---|---|---|
| Kruskal's | Union-Find + Sorting | O(E log E) |
| Prim's | Priority Queue | O(E log V) with a heap |

MST algorithms are useful in network design, infrastructure planning and minimizing connection costs.
