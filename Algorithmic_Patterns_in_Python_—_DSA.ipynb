#An algorithmic pattern is a reusable problem-solving technique that can be applied to multiple coding problems with similar characteristics.

#2. Two Pointers Pattern
The Two Pointers pattern uses two indices to process a data structure without unnecessary nested loops.

It is commonly used with arrays, strings and linked lists.
Problem: Find two numbers whose sum equals a target

```python
arr = [1, 2, 4, 6, 8, 9]

target = 10
```

Brute Force Approach

```python
def two_sum(arr, target):

    for i in range(len(arr)):

        for j in range(i + 1, len(arr)):

            if arr[i] + arr[j] == target:
                return i, j

    return -1, -1
```

Time complexity: O(n²).

Optimized Two Pointers Approach

```python
def two_sum(arr, target):

    left = 0
    right = len(arr) - 1

    while left < right:

        total = arr[left] + arr[right]

        if total == target:
            return left, right

        elif total < target:
            left += 1

        else:
            right -= 1

    return -1, -1


print(two_sum([1, 2, 4, 6, 8, 9], 10))
```

Time complexity: O(n)
Auxiliary space: O(1)

This particular implementation requires a sorted array.

When to recognize this pattern: Sorted arrays, pair sums, palindrome checking, reversing arrays and removing duplicates.

#3. Sliding Window Pattern
Sliding Window processes a continuous portion of an array or string without recalculating everything from scratch.

There are two types: fixed-size windows and variable-size windows.
##3.1 Fixed-Size Sliding Window
Problem: Find the maximum sum of three consecutive elements

```python
arr = [2, 1, 5, 1, 3, 2]

k = 3
```

[2, 1, 5] 1, 3, 2     Sum = 8

2 [1, 5, 1] 3, 2      Sum = 7

2, 1 [5, 1, 3] 2      Sum = 9

2, 1, 5 [1, 3, 2]     Sum = 6

```python
def max_window_sum(arr, k):

    if not 1 <= k <= len(arr):
        raise ValueError("Invalid window size")

    window_sum = sum(arr[i] for i in range(k))

    max_sum = window_sum

    for right in range(k, len(arr)):

        window_sum += arr[right]

        window_sum -= arr[right - k]

        max_sum = max(max_sum, window_sum)

    return max_sum


print(max_window_sum([2, 1, 5, 1, 3, 2], 3))
```

Time complexity: O(n)
Auxiliary space: O(1)
##3.2 Variable-Size Sliding Window
Problem: Find the longest substring without repeating characters.

```python
def longest_substring(s):

    visited = set()

    left = 0
    max_length = 0

    for right in range(len(s)):

        while s[right] in visited:

            visited.remove(s[left])

            left += 1

        visited.add(s[right])

        max_length = max(
            max_length,
            right - left + 1
        )

    return max_length


print(longest_substring("abcabcbb"))
```

Time complexity: O(n) on average
Auxiliary space: O(min(n, alphabet size))

When to recognize this pattern: Longest substring, shortest subarray, consecutive elements, maximum window sum and moving averages.

#4. Fast and Slow Pointers
This pattern uses two pointers moving at different speeds.

Typically, the slow pointer moves one step while the fast pointer moves two steps.

It is also known as the Floyd's Tortoise and Hare algorithm when used for cycle detection.
Problem: Detect a cycle in a linked list

1 → 2 → 3 → 4 → 5
        ↑       |
        └───────┘

```python
def has_cycle(head):

    slow = head
    fast = head

    while fast and fast.next:

        slow = slow.next

        fast = fast.next.next

        if slow is fast:
            return True

    return False
```

If a cycle exists, the fast pointer eventually meets the slow pointer.

Time complexity: O(n)
Auxiliary space: O(1)

Common problems include detecting linked-list cycles, finding the middle node and identifying the beginning of a cycle.

#5. Prefix Sum Pattern
Prefix Sum stores cumulative sums so that range-sum queries can be answered efficiently.

Consider:

```python
arr = [2, 4, 6, 8, 10]
```

Prefix sum:

Array:    [2, 4,  6,  8, 10]

Prefix: [0, 2, 6, 12, 20, 30]

To calculate the sum from index 1 to index 3:

4 + 6 + 8 = 18

prefix[4] - prefix[1]

20 - 2 = 18


```python
class PrefixSum:

    def __init__(self, arr):

        self.prefix = [0]

        for num in arr:

            self.prefix.append(
                self.prefix[-1] + num
            )

    def range_sum(self, left, right):

        return (
            self.prefix[right + 1]
            - self.prefix[left]
        )


ps = PrefixSum([2, 4, 6, 8, 10])

print(ps.range_sum(1, 3))
```

| Operation | Complexity |
|---|---|
| Preprocessing | O(n) |
| Range query | O(1) |
| Storage | O(n) |

This implementation assumes valid indices and an unchanged original array.

Python also provides accumulate().


```python
from itertools import accumulate

arr = [2, 4, 6, 8, 10]

prefix = list(accumulate(arr, initial=0))

print(prefix)
```

#11. Dynamic Programming Pattern
Dynamic Programming (DP) solves problems by storing solutions to overlapping subproblems instead of calculating them repeatedly.

Two important characteristics are:
- Overlapping subproblems: The same smaller problems occur repeatedly.
- Optimal substructure: A solution can be constructed from solutions to smaller subproblems.
There are two main approaches.
## 11.1 Top-Down DP — Memoization

```python
from functools import lru_cache


@lru_cache(maxsize=None)
def fibonacci(n):

    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


print(fibonacci(10))
```
