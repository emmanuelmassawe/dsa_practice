#Recursion in Python — Data Structures and Algorithms (DSA)
Recursion is a programming technique in which a function calls itself to solve a smaller version of the same problem until a stopping condition is reached.

Recursion is particularly important in DSA because it is used in trees, graphs, backtracking, dynamic programming, and divide-and-conquer algorithms.
#1. Components of Recursion
Every recursive function needs three important components:
- Base Case: The condition that stops recursion.
- Recursive Case: The part where the function calls itself.
- Progress toward the Base Case: Each recursive call must move closer to the stopping condition.
Basic Example

```python
def countdown(n):

    # Base case
    if n <= 0:
        return

    print(n)

    # Recursive case
    countdown(n - 1)


countdown(5)
```

Execution flow:

countdown(5)
    ↓
countdown(4)
    ↓
countdown(3)
    ↓
countdown(2)
    ↓
countdown(1)
    ↓
countdown(0)
    ↓
STOP

Without a base case, the function may continue calling itself until Python raises a RecursionError.
#2. Understanding the Call Stack
To understand recursion properly, you must understand the Call Stack.

Remember the Stack data structure we discussed earlier, which follows LIFO (Last In, First Out).

When a function calls itself, Python creates a new stack frame containing information such as local variables and where execution should resume.

Consider:

```python
def factorial(n):

    if n <= 1:
        return 1

    return n * factorial(n - 1)


print(factorial(4))
```

```python
def count(n):
  if n==0:
    return 0
  else:

    count(n-1)

count(400)
```

4. Types of Recursion.

| Type | Description |
|---|---|
| Direct Recursion | A function calls itself directly |
| Indirect Recursion | Functions call each other recursively |
| Linear Recursion | Each non-base invocation makes one recursive call |
| Binary Recursion | An invocation can make two recursive calls |
| Multiple Recursion | An invocation can make more than two recursive calls |
| Tail Recursion | The recursive call is the final operation |
| Non-tail Recursion | Additional operations occur after the recursive call returns |

#5. Common Recursion Algorithms

## 5.1 Sum of Natural Numbers
Problem: Calculate the sum of numbers from 1 to n.

```python
def sum_numbers(n):

    if n == 0:
        return 0

    return n + sum_numbers(n - 1)


print(sum_numbers(5))
# 15
```

Time Complexity: O(n)

Space Complexity: O(n)
## 5.2 Sum of Array Elements

```python
def array_sum(arr, index=0):

    if index == len(arr):
        return 0

    return arr[index] + array_sum(arr, index + 1)


arr = [10, 20, 30, 40]

print(array_sum(arr))
# 100
```

Time Complexity: O(n)

Space Complexity: O(n)

Using an index avoids creating a new array during every recursive call.
# 5.3 Reverse a String

```python
def reverse_string(s):

    if len(s) <= 1:
        return s

    return reverse_string(s[1:]) + s[0]


print(reverse_string("Python"))
# nohtyP
```

Time Complexity: O(n²)

Space Complexity: O(n²) in this implementation because recursive calls retain sliced strings.

For practical Python applications, s[::-1] is more efficient.
##5.4 Check Palindrome
This combines Recursion and Two Pointers.

```python
def is_palindrome(s, left, right):

    if left >= right:
        return True

    if s[left] != s[right]:
        return False

    return is_palindrome(
        s,
        left + 1,
        right - 1
    )


text = "racecar"

print(is_palindrome(text, 0, len(text) - 1))
# True
```

Time Complexity: O(n)

Space Complexity: O(n)
#6. Fibonacci Sequence and Recursion Tree
The Fibonacci sequence is defined as:
\[
F(n) = F(n-1) + F(n-2)
\]
Where:
\[
F(0)=0,\qquad F(1)=1
\]
Sequence:

```python
def fibonacci(n):

    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


print(fibonacci(5))
# 5
```

Recursion Tree

                    fib(5)
                  /        \
              fib(4)       fib(3)
              /    \       /    \
          fib(3) fib(2) fib(2) fib(1)
          /   \   /  \   /  \
      fib(2) fib(1)  ...     ...
       /  \
   fib(1) fib(0)

Notice that fib(3) and fib(2) are calculated multiple times.

This is called Overlapping Subproblems.

Time Complexity: O(2ⁿ) upper bound.

Space Complexity: O(n) because the maximum call stack depth grows linearly.
Optimizing Recursion Using Memoization
Memoization stores previously calculated results so that they are not calculated again.

```python
def fibonacci(n, memo=None):

    if memo is None:
        memo = {}

    if n <= 1:
        return n

    if n in memo:
        return memo[n]

    memo[n] = (
        fibonacci(n - 1, memo)
        + fibonacci(n - 2, memo)
    )

    return memo[n]


print(fibonacci(10))
# 55
```

Alternatively, Python provides lru_cache

```python
from functools import lru_cache


@lru_cache(maxsize=None)
def fibonacci(n):

    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


print(fibonacci(10))
# 55
```

Performance Comparison

| Implementation | Time Complexity | Space Complexity |
|---|---|---|
| Naive recursion | O(2ⁿ) upper bound | O(n) |
| Recursion + Memoization | O(n) | O(n) |
| Iterative approach | O(n) | O(1) |

#11. How to Calculate Recursion Time Complexity
You can analyze recursive algorithms using recurrence relations.

| Algorithm | Recurrence | Time Complexity |
|---|---|---|
| Factorial | T(n) = T(n−1) + O(1) | O(n) |
| Array Sum | T(n) = T(n−1) + O(1) | O(n) |
| Binary Search | T(n) = T(n/2) + O(1) | O(log n) |
| Merge Sort | T(n) = 2T(n/2) + O(n) | O(n log n) |
| Naive Fibonacci | T(n) = T(n−1) + T(n−2) + O(1) | O(2ⁿ) upper bound |
| Generate Subsets | T(n) = 2T(n−1) + O(n) upper bound | O(n × 2ⁿ) upper bound |



13. Recursion Limit in Python
Python has a recursion depth limit to protect against excessive stack usage.

```python
import sys

print(sys.getrecursionlimit())
```

```python
sys.setrecursionlimit(2000)
```

However, increasing the recursion limit unnecessarily can cause excessive memory consumption or interpreter crashes.

For production systems processing large graphs or deeply nested structures, an iterative implementation using an explicit stack is often safer.

#14. Common Recursion Mistakes

| Mistake | Consequence |
|---|---|
| Missing base case | RecursionError |
| Not progressing toward base case | Infinite recursion |
| Forgetting `return` | Incorrect result or `None` |
| Recalculating subproblems | Poor performance |
| Creating unnecessary slices | Increased memory and time complexity |
| Excessive recursion depth | RecursionError |
| Missing visited set in cyclic graphs | Repeated traversal |
| Forgetting to undo backtracking choices | Incorrect solutions |
| Using mutable default arguments | Unexpected shared state |
